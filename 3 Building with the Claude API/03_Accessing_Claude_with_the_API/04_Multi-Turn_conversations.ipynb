```python
# Load env variables and create client
import os
import debugpy
import json
from dotenv import load_dotenv
from anthropic import Anthropic
from IPython.display import display
import pprint, shutil
import textwrap

load_dotenv()
if not os.getenv("ANTHROPIC_API_KEY"):
    exit("No API key found in environment variables")

client = Anthropic()
model = "claude-haiku-4-5"

def myprint(messages):
    width = shutil.get_terminal_size().columns
    for line in json.dumps(messages, indent=2, ensure_ascii=False).splitlines():
        indent = len(line) - len(line.lstrip())
        print(textwrap.fill(line, width=width, subsequent_indent=' ' * (indent + 2)))
```

```python
message = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=[
        {
            "role": "user",
            "content": "What is quantum computing? Answer in one sentence"
        }
    ]
)
```

```python
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages):
    message = client.messages.create(
        model=model,
        max_tokens=1000,
        messages=messages,
    )
    return message.content[0].text
```

```python
# Start with an empty message list
messages = []

# Add the initial user question
add_user_message(messages, "Define quantum computing in one sentence")

# Get Claude's response
answer = chat(messages)
debugpy.wait_for_client

# Add Claude's response to the conversation history
add_assistant_message(messages, answer)

# Add a follow-up question
add_user_message(messages, "Write another sentence")

# Get the follow-up response with full context
final_answer = chat(messages)
# breakpoint()  # execution pauses here
# from IPython.core.debugger import set_trace
# set_trace()  # drops into an ipdb session inline in the cell output
```

```python
print("messages:")
myprint(messages)

print("final answer:")
myprint(final_answer)
```
